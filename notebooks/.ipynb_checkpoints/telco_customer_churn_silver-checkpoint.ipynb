# Limpeza e tratamento dos dados

## Carregando os dados

```python
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)

DATA_PATH = '../data/'
file_name = 'telco_customer_churn_raw.csv'
df_raw_data = pd.read_csv(DATA_PATH+file_name, sep=',')
df_raw_data.convert_dtypes()
df_raw_data.shape
```

## Conhecendo as variáveis

```python
df_raw_data.head(5)
```

```python
df_raw_data.dtypes
```

## Identificando valores duplicados e nulos

```python
# identificando valores duplicados

df_raw_data.duplicated().sum()
```

```python
# identifica valores duplicados na coluna customerID

df_raw_data.duplicated(subset=['customerID']).sum()
```

```python
# identificando valores nulos

df_raw_data.isna().sum()
```

```python
# identificando espaços em strings

for column in df_raw_data.columns:
    print(f'{column}: {df_raw_data[column].replace(r'^\s*$', pd.NA, regex=True).isna().sum()}')
```

```python
# Verifica se os valores fazem sentido para a coluna
for column in df_raw_data.columns:
    print(f'{column}: {df_raw_data[column].unique()}')
```

## Tratando valores nulos identificados na coluna TotalCharges

```python
df_raw_data[df_raw_data['TotalCharges'].replace(r'^\s*$', pd.NA, regex=True).isna() == True].shape
```

```python
# Separa dataframe sem valores nulos

df_raw_data = df_raw_data[df_raw_data['TotalCharges'].replace(r'^\s*$', pd.NA, regex=True).isna() == False]
df_raw_data.shape
```

```python
# verifica novamente existencia de valores vazios

for column in df_raw_data.columns:
    print(f'{column}: {df_raw_data[column].replace(r'^\s*$', pd.NA, regex=True).isna().sum()}')
```

```python
df_raw_data['TotalCharges'] = df_raw_data['TotalCharges'].astype(float)
df_raw_data.dtypes
```

## Altera SeniorCitizen para Yes/No

```python
df_raw_data['SeniorCitizen'] = df_raw_data['SeniorCitizen'].astype(object)
df_raw_data['SeniorCitizen'] = df_raw_data['SeniorCitizen'].replace(0,'No')
df_raw_data['SeniorCitizen'] = df_raw_data['SeniorCitizen'].replace(1,'Yes')
df_raw_data['SeniorCitizen']
```

## Renomeando colunas

```python
df_raw_data = df_raw_data.rename({'customerID':'CustomerID', 'gender':'Gender', 'tenure':'Tenure'}, axis=1)
df_raw_data.head(1)
```

## Salva o dataset limpo

```python
file_name = 'telco_customer_churn_silver.csv'
df_raw_data.to_csv(DATA_PATH+file_name, sep=',', index=None)
```
